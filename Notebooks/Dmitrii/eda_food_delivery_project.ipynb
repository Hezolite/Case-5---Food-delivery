### 3.2 Анализ выбросов и физической адекватности данных

```python
import matplotlib.pyplot as plt
import seaborn as sns

num_features = [
    'actual_delivery_time_minutes',
    'restaurant_preparation_time_minutes',
    'distance_km',
    'delivery_partner_experience_months',
    'delivery_partner_rating',
    'subtotal'
]

fig, axes = plt.subplots(2, 3, figsize=(16, 8))
axes = axes.flatten()

for idx, col in enumerate(num_features):
    sns.boxplot(x=df_delivered[col], ax=axes[idx], color='skyblue')
    axes[idx].set_title(col, fontsize=11, fontweight='bold')
    axes[idx].set_xlabel('')

plt.suptitle('Поиск выбросов в ключевых признаках', fontsize=15, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```

В принципе, данные нормальные, выбросы конечно есть, но они ложатся в реальную картину мира. Например, дистанция реально может быть 20 километров (например, коттеджный поселок за городом), а время готовки уйти за 70 минут, если первое приготовление не удастся. То же самое с временем доставки, курьер может попасть в сильную пробку или в аварию, тут нет запредельных значений, например, заказ за 1000 км или время доставки 3 года. Поэтому выбросы не удаляем, иначе модель не научиться предсказывать сложные далеки заказы в час пик. 

Нужно еще проверить нет ли странных выбросов со скоростью доставки, вдруг там курьеры летают как метеоры

```python
transit_time = df_delivered['actual_delivery_time_minutes'] - df_delivered['restaurant_preparation_time_minutes']

print("Минимальное время в пути (мин):", transit_time.min())
print("Заказов, где время доставки <= времени готовки:", (transit_time <= 0).sum())

valid_mask = transit_time > 0
speed_kmh = df_delivered.loc[valid_mask, 'distance_km'] / (transit_time[valid_mask] / 60)

print("\nСкорость курьера в пути (км/ч):")
print(speed_kmh.describe().round(2))
```

Тут, конечно, ситуация поинтереснее вырисовывается. Если попробовать объяснить логически, то время в пути -5 минут может быть в том случае, если по нормативу блюдо готовится, например, 10 минут, но мы приготовили часть ингридиентов заранее и отдали в доставку за 5 минут. Ну либо же также шум во время генерации

```python
transit_time = df_delivered['actual_delivery_time_minutes'] - df_delivered['restaurant_preparation_time_minutes']
speed_kmh = df_delivered['distance_km'] / (transit_time / 60)

df_clean = df_delivered[(transit_time > 0) & (speed_kmh <= 60)].copy()

print(f"Удалено аномалий: {len(df_delivered) - len(df_clean)} ({(1 - len(df_clean) / len(df_delivered)) * 100:.2f}%)")
print(f"Итоговый размер чистой выборки df_clean: {len(df_clean)} строк")
```

Удалили 0.44% строк с физически нереальным временем в пути <= 0 или скоростью > 60 км/ч

## Часть 4. Одномерный анализ

### 4.1 Распределение целевой переменной и сравнение с прогнозом платформы

```python
fig, axes = plt.subplots(1, 2, figsize=(15, 5))

target = df_clean['actual_delivery_time_minutes']
sns.histplot(target, bins=50, kde=True, ax=axes[0], color='steelblue')
axes[0].axvline(target.mean(), color='red', linestyle='--', linewidth=2, label=f'Mean: {target.mean():.1f} мин')
axes[0].axvline(target.median(), color='green', linestyle='--', linewidth=2, label=f'Median: {target.median():.1f} мин')
axes[0].set_title('Распределение фактического времени доставки', fontsize=13, fontweight='bold')
axes[0].set_xlabel('Минуты')
axes[0].set_ylabel('Количество заказов')
axes[0].legend()

sns.kdeplot(df_clean['actual_delivery_time_minutes'], ax=axes[1], fill=True, color='steelblue', label='Actual')
sns.kdeplot(df_clean['estimated_delivery_time_minutes'], ax=axes[1], fill=True, color='orange', label='Estimated')
axes[1].set_title('Фактическое время vs Прогноз платформы', fontsize=13, fontweight='bold')
axes[1].set_xlabel('Минуты')
axes[1].set_ylabel('Плотность')
axes[1].legend()

plt.tight_layout()
plt.show()

print(f"Коэффициент асимметрии таргета: {target.skew():.2f}")
```

Быстрее, чем за 10-15 минут привезти заказ нереально, поэтому распределение фактического времени доставки резко обрывается слева. Можно сделать вывод, что текущая модель для оценки очень шаблонная и пытается все заказы вписать в стандартные 20-35 минут, хотя в реальности разброс времени шире. Также можно увидеть, что справа синий график выше, чем оранжевый, следовательно, **текущая модель плохо учитывает сложные условия, из-за которых заказ задерживается**.

### 5.2 Матрица корреляций phik

Преподаватель очень хвалил, так что тоже стоит попробовать

```python
import phik

cols_for_corr = [
    'actual_delivery_time_minutes',
    'distance_km',
    'restaurant_preparation_time_minutes',
    'subtotal',
    'items_count',
    'traffic_level',
    'weather',
    'restaurant_type',
    'city',
    'order_hour',
    'is_weekend',
    'delivery_partner_experience_months',
    'delivery_partner_rating',
    'restaurant_rating'
]

interval_cols = [
    'actual_delivery_time_minutes',
    'distance_km',
    'restaurant_preparation_time_minutes',
    'subtotal',
    'items_count',
    'delivery_partner_experience_months',
    'delivery_partner_rating',
    'restaurant_rating'
]

phik_corr = df_clean[cols_for_corr].phik_matrix(interval_cols=interval_cols)

plt.figure(figsize=(12, 9))
sns.heatmap(
    phik_corr,
    annot=True,
    fmt='.2f',
    cmap='Blues',
    vmin=0,
    vmax=1,
    square=True,
    linewidths=0.5,
    cbar_kws={'shrink': 0.8}
)
plt.title('Матрица нелинейных корреляций φk (phik)', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

Интересно, что есть сильная корелляция между суммой заказа и временем приготовления. Это можно объяснить тем, что чем выше сумма чека, тем дольше кухня будет собирать заказ. Также высокая связт рейтингом ресторана и городом, возможно, связано с тем, что в разных городах доминируют разные типы заведений или локальные сети какие-нибудь

Тут также видно, что опыт и рейтинг курьера никак не влияют на время доставки (0.00 и 0.02 очень маленькие цифры).

## Часть 6. Сегментный анализ и выводы для бизнеса

### 6.1 Влияние погоды и пробок на время доставки и ошибку текущего прогноза

```python
weather_order = ['Clear', 'Cloudy', 'Rain', 'Storm']
traffic_order = ['Low', 'Medium', 'High']

df_seg = df_clean.assign(
    pred_error=df_clean['actual_delivery_time_minutes'] - df_clean['estimated_delivery_time_minutes']
)

piv_actual = df_clean.pivot_table(
    index='weather', columns='traffic_level',
    values='actual_delivery_time_minutes', aggfunc='median', observed=False
).loc[weather_order, traffic_order]

piv_error = df_seg.pivot_table(
    index='weather', columns='traffic_level',
    values='pred_error', aggfunc='mean', observed=False
).loc[weather_order, traffic_order]

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.heatmap(piv_actual, annot=True, fmt='.1f', cmap='YlOrRd', linewidths=0.5, ax=axes[0])
axes[0].set_title('Медианное время доставки (мин)', fontsize=12, fontweight='bold')
axes[0].set_xlabel('Уровень пробок')
axes[0].set_ylabel('Погода')

sns.heatmap(piv_error, annot=True, fmt='+.1f', cmap='Reds', linewidths=0.5, ax=axes[1])
axes[1].set_title('Среднее опоздание прогноза платформы (мин)', fontsize=12, fontweight='bold')
axes[1].set_xlabel('Уровень пробок')
axes[1].set_ylabel('Погода')

plt.tight_layout()
plt.show()
```

Если смотреть на правый график, видно, что при ясной погоде уровень пробок практически никак не влияет на опоздание прогноза, но чем хуже погода, тем сильнее она воздействует. Хорошо, что мы не удалили Storm, несмотря на малое количество записей по сравнению с другой погодой, потому что влияние ощутимое. 

Комбо из шторма и сильных пробок увеличивает реальное время доставки почти в 2 раза: с 28 минут (Clear + Low) до 50 минут (Storm + High).

Можно сделать вывод, что текущая формула платформы умеет учитывать пробки в ясную погоду (ошибка ровная, около +2.5 мин), но проваливается на плохой погоде, занижая прогноз на от 10 до 18.6 минут. Для нашей будущей ML-модели это пригодится.

### 6.2 Портрет проблемной (долгой) доставки

```python
import numpy as np

q90 = df_clean['actual_delivery_time_minutes'].quantile(0.90)

df_prof = pd.DataFrame({
    'is_long': df_clean['actual_delivery_time_minutes'] >= q90,
    'Дистанция': df_clean['distance_km'],
    'Время готовки': df_clean['restaurant_preparation_time_minutes'],
    'Сумма чека': df_clean['subtotal'],
    'Кол-во блюд': df_clean['items_count'],
    'Пробки High': (df_clean['traffic_level'] == 'High').astype(float),
    'Дождь / Шторм': df_clean['weather'].isin(['Rain', 'Storm']).astype(float),
    'Опыт курьера': df_clean['delivery_partner_experience_months']
})

features = [
    'Дистанция', 'Время готовки', 'Сумма чека',
    'Кол-во блюд', 'Пробки High', 'Дождь / Шторм', 'Опыт курьера'
]

comp_table = df_prof.groupby('is_long')[features].mean().T
comp_table.columns = [f'Обычные (<{q90:.0f} мин)', f'Долгие (>={q90:.0f} мин)']
comp_table['Разница %'] = ((comp_table.iloc[:, 1] / comp_table.iloc[:, 0] - 1) * 100).round(1)
display(comp_table.round(2))

norm_df = df_prof[features].copy()
for col in ['Дистанция', 'Время готовки', 'Сумма чека', 'Кол-во блюд', 'Опыт курьера']:
    norm_df[col] = (norm_df[col] / norm_df[col].quantile(0.95)).clip(0, 1)
norm_df['is_long'] = df_prof['is_long']

radar_stats = norm_df.groupby('is_long')[features].mean()

angles = np.linspace(0, 2 * np.pi, len(features), endpoint=False).tolist()
angles += angles[:1]
v_long = radar_stats.loc[True].tolist() + [radar_stats.loc[True].tolist()[0]]
v_norm = radar_stats.loc[False].tolist() + [radar_stats.loc[False].tolist()[0]]

fig, ax = plt.subplots(figsize=(8, 8), subplot_kw=dict(polar=True))
ax.plot(angles, v_long, 'o-', linewidth=2.5, label='Долгие доставки (топ-10%)', color='#e74c3c')
ax.fill(angles, v_long, alpha=0.25, color='#e74c3c')
ax.plot(angles, v_norm, 'o-', linewidth=2.5, label='Обычные доставки', color='#3498db')
ax.fill(angles, v_norm, alpha=0.15, color='#3498db')

ax.set_xticks(angles[:-1])
ax.set_xticklabels(features, fontsize=11, fontweight='bold')
ax.set_ylim(0, 1)
ax.legend(loc='upper right', bbox_to_anchor=(1.42, 1.05))
ax.set_title('Профиль долгой доставки vs Обычные заказы', fontsize=13, fontweight='bold', pad=20)
plt.tight_layout()
plt.show()
```

График в третий раз доказывает, что опыт курьера никак не влияет на ситуацию и не создает разницы между обычной и долгой доставкой. А вот дистанция, время готовки, сумма чека, кол-во блюд, сильные пробки и непогода влияют сильно. Причем наибольший вес имеет именно дистанция, что также подтверждает матрица корелляций.

### 1. Что с самими данными
В датасете 50 000 заказов и 37 столбцов, дубликатов нет. После перевода текстовых колонок с повторами в тип `category` и сжатия числовых типов размер таблицы в памяти уменьшился с 53.2 до 7.4 Мб (на 86%).

Случайных пропусков тоже нет: пустые ячейки встречаются только у 868 отмененных заказов (причем почти половина отмен из-за `Driver Unassigned`, то есть банально не хватило свободных курьеров). Для нашей задачи мы оставили только выполненные заказы, а затем убрали еще 218 строк (0.44%), где время в пути получалось меньше нуля или скорость курьера улетала за 60 км/ч. В итоге для обучения осталась чистая выборка `df_clean` на 48 914 строк.

### 2. Что не так с текущим прогнозом платформы
В среднем заказ едет 35 минут (по медиане - 33 минуты), но из-за сложных и далеких заказов у графика есть длинный хвост вправо вплоть до 170 минут.

Старый алгоритм платформы оценивает время слишком оптимистично и шаблонно (медиана прогноза - 28 минут), пытаясь уложить почти все заказы в рамки 20-35 минут. Из-за этого 37% заказов опаздывают. Сильнее всего старая формула ошибается в плохую погоду: если в ясный день среднее опоздание держится около 2.5 минут, то в дождь и шторм с пробками прогноз врет уже на 10-18.6 минут.

### 3. Что реально влияет на время доставки
Сильнее всего время доставки зависит от дистанции (`0.77` по матрице `phik`), времени готовки блюда (`0.56`), суммы чека и количества позиций (`0.48` и `0.41`), пробок (`0.27`), типа ресторана (`0.21`) и погоды (`0.18`). По часам заказа хорошо видны два пика нагрузки: обед (11-13 часов) и ужин (18-21 час). При этом сумма заказа очень тесно связана со временем готовки (`0.93`) - чем больше чек, тем дольше кухня собирает заказ.

Главный сюрприз: опыт курьера (`0.02`) и его рейтинг (`0.00`) вообще никак не влияют на скорость доставки, хотя в условии кейса на них делался акцент. А вот информации о транспорте курьера (пешком, на велосипеде или на машине) в датасете, к сожалению, нет.

### 4. Что убираем и что передаем дальше на этап обучения модели
* Не берем в модель колонки, которые появляются только после завершения заказа: `late_delivery`, `customer_rating`, `order_status`, `cancellation_reason` и `tip_amount`.
* Колонку `estimated_delivery_time_minutes` не подаем как признак, но сохраняем отдельно - будем сравнивать нашу модель со старым алгоритмом и смотреть, насколько точнее мы стали предсказывать.

* Качество модели лучше замерять через метрику MAE (средняя ошибка в понятных минутах, не так болезненно реагирует на редкие долгие заказы) в паре с RMSE. Для линейных моделей стоит попробовать логарифмировать таргет и не брать одновременно сумму чека и время готовки, чтобы признаки не дублировали друг друга. Из новых фичей имеет смысл добавить флаг часа пик (`11-13` и `18-21`), а также связки погоды и дистанции с уровнем пробок.
